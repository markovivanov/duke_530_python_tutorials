## 3. Give workout details names

> **Concept · Variables**
>
> A variable is a name that refers to a value. The `=` sign assigns the value on its right to the name on its left. Read `length_of_run_km = 5.4` as “store 5.4 under the name length_of_run_km.”

We include units in measurement names. Here, `km` means kilometers and duration is in minutes. Underscores connect words because variable names cannot contain spaces.


```python
runner_name = "Sam"  # Store the runner's name as text.
length_of_run_km = 5.4  # Store the run distance in kilometers.
duration_minutes = 32  # Store the run duration in whole minutes.
run_completed = True  # Store whether the run is complete.
```

```python
print(runner_name)  # Display the value stored under runner_name.
print(length_of_run_km)  # Display the stored run distance.
print(duration_minutes)  # Display the stored duration.
print(run_completed)  # Display the completion flag.
```

`print(runner_name)` looks up the value of a variable. `print("runner_name")` displays the literal text between the quotes. Try predicting the difference before running this cell.


```python
print(runner_name)  # Look up the variable and display Sam.
print("runner_name")  # Display these exact characters as text.
```

## 4. Recognize the kind of value

> **Concept · Data types**
>
> A data type tells Python what kind of value it is working with. These four types let our app represent names, measurements, counts, and yes/no information.

| Type | Meaning | Fitness example |
|---|---|---|
| `str` | Text, written in quotes | `"Sam"` |
| `float` | A number that can have a fractional part | `5.4` kilometers |
| `int` | A whole number | `32` minutes |
| `bool` | Either `True` or `False` | Whether a run is complete |

`type()` reports the type of a value. Below, Python first evaluates the inner `type(...)`, then `print(...)` displays the result. Output such as `<class 'float'>` means the value has type `float`; we will explore the word “class” in week 3.


```python
print(type(runner_name))  # Show that the runner's name is a string.
print(type(length_of_run_km))  # Show that the run distance is a float.
print(type(duration_minutes))  # Show that the duration is an integer.
print(type(run_completed))  # Show that the completion flag is a Boolean.
```

> **Watch out · Quotes change the meaning**
>
> `5.4` is a number, while `"5.4"` is text. `True` and `False` start with capital letters and have no quotes when used as Boolean values.

## 5. Calculate with workout data

> **Concept · Arithmetic**
>
> `+` adds, `-` subtracts, `*` multiplies, and `/` divides. We can use variable names in calculations just as we use numbers. Python calculates the right side of `=` before storing the result.


```python
morning_run_km = 5.4  # Store the distance of a morning run.
evening_run_km = 2.1  # Store the distance of an evening run.
total_distance_km = morning_run_km + evening_run_km  # Add the two distances.
print("Total distance:", total_distance_km, "km")  # Display a label, the total, and its unit.
```

The commas let `print()` display several items with spaces between them. The total is **7.5 km**.


```python
duration_seconds = duration_minutes * 60  # Convert the run's minutes to seconds.
print("Run duration:", duration_seconds, "seconds")  # Display the converted duration.
```

## 6. Build a small run summary

We can now combine the stored details into a readable record.


```python
print("Runner:", runner_name)  # Display the runner's name with a label.
print("Distance:", length_of_run_km, "km")  # Display the run distance and its unit.
print("Duration:", duration_minutes, "minutes")  # Display the duration and its unit.
print("Completed:", run_completed)  # Display whether the run is complete.
```

## 7. Change a value and rerun a calculation

> **Concept · Notebook memory**
>
> The **kernel** is the Python process that runs the code and remembers variables. Changing an input does not automatically update an earlier calculation. Run the calculation again to store a new result.

Predict both numbers below. The first is the total already stored; the second is the newly calculated total.


```python
evening_run_km = 3.0  # Replace the evening distance with a new value.
print("Previously stored total:", total_distance_km, "km")  # Display the old total of 7.5 km.
total_distance_km = morning_run_km + evening_run_km  # Recalculate using the new evening distance.
print("Updated total:", total_distance_km, "km")  # Display the updated total of 8.4 km.
```

> **Watch out · Names and execution order**
>
> Python treats `runner_name` and `Runner_name` as different names. A `NameError` often means a name was misspelled or the cell that creates it has not run. Read the last line of the error, check the spelling, and run the defining cell first.

To return to a predictable starting point, use your editor's **Restart Kernel and Run All** action, or restart the runtime and run all cells. This clears remembered values and rebuilds them in order.
