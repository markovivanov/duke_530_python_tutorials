## 1. Bring in a charting library

> **Concept · Libraries, modules, and import**
>
> A library provides reusable code for a set of tasks. A module is an importable unit of Python code. **Matplotlib** is a library for making charts; its `pyplot` module provides the plotting functions we will use.

`import matplotlib.pyplot as plt` makes that module available under the shorter name `plt`. The name after `as` is an **alias**. We will write calls such as `plt.bar(...)` to use the imported functions.


```python
import matplotlib.pyplot as plt  # Import Matplotlib's plotting module under the short name plt.
```

Importing makes the plotting functions available. It does not draw a chart yet.

## 2. Prepare the chart's data

> **Concept · Match each label with a value**
>
> Our first list names the days in order. Our second list contains the distance for each corresponding day. The first label matches the first distance, and so on. The two lists must have the same number of items.


```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]  # Store the chart's day labels in order.
daily_distances_km = [3.0, 0.0, 5.0, 2.0, 0.0, 4.0, 6.0]  # Store the matching recorded distances.
print("Day labels:", len(days))  # Display the number of labels.
print("Distance values:", len(daily_distances_km))  # Display the number of values.
```

Both counts are seven. Tuesday and Friday have zero recorded run distance, so their bars will have zero height. This chart describes recorded runs only.

## 3. Draw a first bar chart

> **Concept · Start, draw, label, show**
>
> `plt.figure()` starts a new figure, `plt.bar()` draws bars, and `plt.show()` displays the result. The x-axis runs horizontally and names the days. The y-axis runs vertically and measures distance.


```python
plt.figure()  # Start a fresh figure for this chart.
plt.bar(days, daily_distances_km)  # Draw one bar per day using the corresponding distance.
plt.title("Recorded run distance this week")  # Add a title explaining the chart's subject.
plt.xlabel("Day")  # Label the horizontal axis.
plt.ylabel("Distance (km)")  # Label the vertical axis with the measured unit.
plt.show()  # Display the completed chart in the notebook.
```

### Read the chart together

- Which day has the largest recorded distance?
- Which days have no recorded run?
- What does the vertical axis measure?

The tallest bar is Sunday at **6.0 km**. A chart is useful only if its labels make the values understandable.

## 4. Make a small presentation change

> **Concept · Keyword arguments**
>
> Some functions accept optional settings. A keyword argument names the setting, followed by `=` and a value. Here, `color="teal"` chooses the bar color. It is an argument to `plt.bar()`, rather than a separate variable assignment.

We keep a zero baseline for these distance bars so their heights can be compared directly. `plt.ylim(bottom=0)` sets the lower end of the vertical axis; `plt.tight_layout()` adjusts spacing around the labels.


```python
plt.figure()  # Start a separate figure for the styled version.
plt.bar(days, daily_distances_km, color="teal")  # Draw the distances using a named bar color.
plt.title("Sam's recorded runs this week")  # Give the chart a specific descriptive title.
plt.xlabel("Day")  # Identify the horizontal labels as days.
plt.ylabel("Recorded distance (km)")  # State what each bar measures and its unit.
plt.ylim(bottom=0)  # Set the distance axis to start at zero.
plt.tight_layout()  # Adjust spacing so labels fit within the figure.
plt.show()  # Display the updated chart.
```

## 5. Update a recorded run

Our runner adds a 2 km Tuesday run. Recall that index `1` refers to the second list item. Predict the changed bar before running these cells.


```python
daily_distances_km[1] = 2.0  # Update Tuesday's recorded distance to 2 km.
print("Updated distances:", daily_distances_km)  # Inspect the changed list before plotting.
```

```python
plt.figure()  # Start a fresh figure using the updated data.
plt.bar(days, daily_distances_km, color="teal")  # Draw bars from the current distance values.
plt.title("Sam's recorded runs after Tuesday's update")  # Explain which version of the data is shown.
plt.xlabel("Day")  # Label the horizontal axis.
plt.ylabel("Recorded distance (km)")  # Label the vertical axis with units.
plt.ylim(bottom=0)  # Keep the bar chart's baseline at zero.
plt.tight_layout()  # Fit the labels within the figure.
plt.show()  # Display the newly generated chart.
```

The earlier chart is a saved output from an earlier execution. Updating the list does not redraw it automatically; running the plotting cell builds a new result.

## 6. Connect the chart to a number

Recall the total from week 6: start at zero and add each distance in a loop. We can show that total beside the visual summary.


```python
weekly_total_km = 0.0  # Start a fresh total for the current chart data.
for distance_km in daily_distances_km:  # Visit each distance shown in the latest chart.
    weekly_total_km = weekly_total_km + distance_km  # Add this day's distance to the total.
print("Distance represented in the updated chart:", weekly_total_km, "km")  # Display the updated total of 22.0 km.
```
