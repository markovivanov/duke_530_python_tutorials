Recall that lists contain items, loops process them in order, and an accumulator starts at zero. Keep this notebook beside the supplied data folder. This lesson reads data/run_distances_km.txt; it includes all seven sample distances and needs no earlier notebook.

## 1. Find the supplied workout log

> **Concept · A file and its path**
>
> A file stores information on disk. A **path** describes where to find it. The relative path `data/run_distances_km.txt` means “inside the data folder under Python's current working folder, find run_distances_km.txt.”

Start Jupyter from the workshop folder, or open the notebook in that folder in your editor. The folder layout should include:

```text

```

If using Colab, create a `data` folder in its Files panel and upload the supplied text file into it before running this lesson. The file has one distance per line: 3.0, 0.0, 5.0, 2.0, 0.0, 4.0, and 6.0 kilometers, in Monday-to-Sunday order.


````python
input_path = "data/run_distances_km.txt"  # Store the relative path to the supplied workout log.
print("Workout log:", input_path)  # Display the path we will use.
````

## 2. Open the file and read its text

> **Concept · with open(...)**
>
> `open()` opens a file. The mode `"r"` means read. `as log_file` gives the opened file a name we can use inside the indented block. `with` closes the file automatically when that block finishes.

`encoding="utf-8"` names the text encoding used to interpret the file's characters. This is a **keyword argument**: the input name is written explicitly before `=`. The `.read()` method retrieves the file's contents as one string.


````python
with open(input_path, "r", encoding="utf-8") as log_file:  # Open the sample log for reading text.
    log_text = log_file.read()  # Read the complete file into one string.
print(log_text)  # Display the saved distances on separate lines.
````

The digits came from a text file, so they are currently text. `type()` lets us check.


````python
print(type(log_text))  # Confirm that reading the file produced a string.
````

## 3. Turn numeric text into a number

> **Concept · Type conversion**
>
> `float()` converts suitable numeric text into a decimal number. This lets us calculate with a saved distance. In the other direction, `str()` converts a value to text for writing into a text file.


````python
sample_distance_text = "3.0"  # Store a distance as text, as it appears in the file.
sample_distance_km = float(sample_distance_text)  # Convert the text into a numeric distance.
print(type(sample_distance_km))  # Confirm that the converted value has type float.
print("Distance plus 1 km:", sample_distance_km + 1.0)  # Use the converted number in arithmetic.
````

## 4. Read a list of lines

The `.readlines()` method gives us a list of strings, one per line. It keeps the newline character, written as `\n`, where one is present. `float()` accepts the trailing newline when converting a numeric line.


````python
with open(input_path, "r", encoding="utf-8") as log_file:  # Reopen the log to read it from the beginning.
    distance_lines = log_file.readlines()  # Read the file into a list of line strings.
print(distance_lines)  # Display the list, including its newline characters.
print("Days recorded:", len(distance_lines))  # Count the seven saved daily entries.
````

## 5. Calculate the weekly distance

Now combine three familiar actions: loop through the list, convert each line to a number, and add it to a running total.


````python
total_distance_km = 0.0  # Start a fresh total before reading the distance values.
for distance_text in distance_lines:  # Visit each saved daily distance as text.
    distance_km = float(distance_text)  # Convert the current line into a numeric distance.
    total_distance_km = total_distance_km + distance_km  # Add the current distance to the running total.
print("Weekly run distance:", total_distance_km, "km")  # Display the total of 20.0 km.
````

## 6. Prepare a summary as text

> **Concept · Build the text to save**
>
> For strings, `+` joins text together. Convert the numeric total with `str()` before joining it to the labels. A newline character `\n` asks the text to continue on the next line.


````python
total_as_text = str(total_distance_km)  # Convert the numeric total into text.
summary_text = "Weekly run distance: " + total_as_text + " km\n"  # Join a label, the total, units, and a newline.
print(summary_text)  # Preview the summary text before saving it.
````

## 7. Save the summary and read it back

> **Concept · Write mode**
>
> Mode `"w"` creates a file, or replaces its contents if that file already exists. This example writes the workshop output `fitness_weekly_summary.txt` beside the notebook. Running the cell again replaces that generated summary with the current result.

The input log and output summary use different paths. The `.write()` method writes our string into the output file.


````python
output_path = "fitness_weekly_summary.txt"  # Choose a filename for the generated summary.
with open(output_path, "w", encoding="utf-8") as summary_file:  # Open the generated summary in replacement write mode.
    summary_file.write(summary_text)  # Save the prepared summary text to disk.
print("Saved summary to:", output_path)  # Display the name of the generated file.
````

````python
with open(output_path, "r", encoding="utf-8") as summary_file:  # Open the saved summary for a read-back check.
    saved_summary = summary_file.read()  # Retrieve the text that is actually on disk.
print(saved_summary)  # Display the saved summary for comparison with the preview.
````

> - `ValueError` from `float()`: check whether a line contains numeric text. This sample file uses values such as `3.0`, with no `km` suffix or blank lines.
> - `TypeError` when joining strings: convert the numeric total with `str()` before joining it to text.

### Key takeaways

- Text read from a file needs conversion before numerical calculations.
- `with open(...)` manages an opened file, and the mode determines whether we read or write.
