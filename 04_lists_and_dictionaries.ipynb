## 1. Store several runs in one place

Separate variables can describe separate runs, but a workout history is easier to manage as one collection.


```python
monday_run_km = 3.0  # Store the first sample run distance.
wednesday_run_km = 5.0  # Store the second sample run distance.
friday_run_km = 2.0  # Store the third sample run distance.
print(monday_run_km, wednesday_run_km, friday_run_km)  # Display the three separate values.
```

> **Concept · Lists**
>
> A list holds items in order. Square brackets `[]` surround the items, and commas separate them. Our distance list follows the order in which the runs were recorded.


```python
run_distances_km = [3.0, 5.0, 2.0]  # Store the same three distances in one ordered list.
print("Recorded distances:", run_distances_km)  # Display the complete list.
print(type(run_distances_km))  # Confirm that this value has type list.
```

## 2. Retrieve one recorded run

> **Concept · An index is a position**
>
> Python counts list positions from **zero**. Put the index in square brackets after the list name to retrieve an item.

| Index | Position | Distance |
|---|---|---|
| `0` | First recorded run | 3.0 km |
| `1` | Second recorded run | 5.0 km |
| `2` | Third recorded run | 2.0 km |


```python
first_distance_km = run_distances_km[0]  # Retrieve the first item using index zero.
second_distance_km = run_distances_km[1]  # Retrieve the second item using index one.
print("First run:", first_distance_km, "km")  # Display the first run's distance.
print("Second run:", second_distance_km, "km")  # Display the second run's distance.
```

`len()` reports how many items a collection contains. For a list with three items, the valid nonnegative indexes are 0, 1, and 2.


```python
number_of_runs = len(run_distances_km)  # Count how many distances are in the list.
print("Number of recorded runs:", number_of_runs)  # Display a count of three.
```

> **Watch out · An index must exist**
>
> Asking this three-item list for index `3` would raise an `IndexError`. The count is three, but the final index is two because counting starts at zero.

## 3. Add a new run and correct an old one

> **Concept · A list can change**
>
> The `.append()` method adds one item to the end of a list. Assigning to an existing index replaces that item. These operations change the list itself.


```python
run_distances_km.append(4.0)  # Add a new 4 km run to the end of the history.
print("Updated distances:", run_distances_km)  # Display the four recorded distances.
print("Number of runs:", len(run_distances_km))  # Display the new count of four.
```

```python
run_distances_km[1] = 5.5  # Correct the second run's distance to 5.5 km.
print("Corrected distances:", run_distances_km)  # Display the list after the correction.
```

> **Notebook habit · Repeating a change repeats its effect**
>
> Running the `.append()` cell again adds another item. To restore the original history, rerun the cell that assigns `[3.0, 5.0, 2.0]`, then run the later cells once in order.

## 4. Describe one workout with named details

The list records distances, but one workout also has an activity and duration. Named fields help us retrieve each detail clearly.

> **Concept · Dictionaries**
>
> A dictionary holds **key–value pairs**. Curly braces `{}` surround the pairs. Each colon connects a key to its value, and commas separate pairs. Our keys are text labels such as `"activity"` and `"distance_km"`.


```python
workout = {  # Create a dictionary describing one recorded workout.
    "activity": "run",  # Store the activity name under the activity key.
    "distance_km": 5.4,  # Store a numeric distance under the distance_km key.
    "duration_minutes": 32,  # Store a numeric duration under the duration_minutes key.
}
print(workout)  # Display the dictionary and its named values.
```

Use square brackets to retrieve a value, just as with a list. The lookup inside the brackets is a **key** here, rather than a list position.


```python
print("Activity:", workout["activity"])  # Retrieve the workout's activity by its key.
print("Distance:", workout["distance_km"], "km")  # Retrieve the distance by its key.
print("Duration:", workout["duration_minutes"], "minutes")  # Retrieve the duration by its key.
```

## 5. Update the workout record

Assigning to an existing key changes its value. Assigning to a new key adds that field to the dictionary.


```python
workout["duration_minutes"] = 33  # Correct the stored duration to 33 minutes.
workout["completed"] = True  # Add a completion flag as a new field.
print("Updated workout:", workout)  # Display the changed and newly added fields.
```

## 6. Build a workout summary

Our app can retrieve measurements from the dictionary and calculate pace. Recall that duration divided by distance gives minutes per kilometer, and `round(value, 2)` rounds for display.


```python
distance_km = workout["distance_km"]  # Read the recorded distance into a named variable.
duration_minutes = workout["duration_minutes"]  # Read the corrected duration into a named variable.
pace = duration_minutes / distance_km  # Calculate pace using the workout's values.
print("Workout:", workout["activity"])  # Display the activity label.
print("Pace:", round(pace, 2), "min/km")  # Display the pace of 6.11 decimal minutes per kilometer.
```

### Key takeaways

- Lists group values that we access by position.
- Dictionaries group values that we access by a named key.
