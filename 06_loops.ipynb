### Getting started

Recall that a list stores items in order, a comparison produces True or False, and an if statement runs an indented body when its condition is true. We supply a fresh seven-day list below; zero means no run was recorded that day.

All people and activity records in this notebook are fictional. Run code cells from top to bottom. Each notebook includes its own examples and starter definitions.


## 1. Meet the week's data

The list below contains Monday through Sunday's recorded distances. A zero represents a day with no recorded run; it does not describe any other activity that day.


```python
daily_distances_km = [3.0, 0.0, 5.0, 2.0, 0.0, 4.0, 6.0]  # Store seven daily run distances in order.
print("Daily run distances:", daily_distances_km)  # Display the supplied week's data.
print("Days recorded:", len(daily_distances_km))  # Confirm that the list contains seven days.
```

## 2. Visit each day with a for loop

> **Concept · Repeating an instruction**
>
> A `for` loop takes one item at a time from a collection. Here, `distance_km` receives the next distance from the list, and the indented body runs using that value. The loop stops after the final item.

The word `in` connects the temporary loop variable to the collection. `distance_km` is a name we choose; `daily_distances_km` is our existing list.


```python
for distance_km in daily_distances_km:  # Visit each recorded daily distance in order.
    print("Recorded run distance:", distance_km, "km")  # Display the current day's distance.
```

You should see seven lines of output, one per list item. The print instruction is written once, but the loop runs it seven times.

## 3. Build a total one item at a time

> **Concept · An accumulator**
>
> An accumulator is a variable that stores a running total. Start it at zero before the loop. On each repetition, add the current distance to the value already stored.

`total_km = total_km + distance_km` reads the old total, adds the current distance, and stores the new total under the same name.


```python
total_km = 0.0  # Start a new running total before processing the week.
for distance_km in daily_distances_km:  # Visit each day's run distance.
    total_km = total_km + distance_km  # Add the current distance to the previous total.
    print("Total so far:", total_km, "km")  # Display the running total after this addition.
print("Weekly total:", total_km, "km")  # Display the final total after the loop ends.
```

### Trace the first three repetitions

| Current distance | Total before adding | Total after adding |
|---|---|---|
| 3.0 km | 0.0 km | 3.0 km |
| 0.0 km | 3.0 km | 3.0 km |
| 5.0 km | 3.0 km | 8.0 km |

The full week's total is **20.0 km**. The final print is outside the loop, so it runs once.

> **Watch out · Reset before repeating the whole calculation**
>
> The total must start at zero before the loop. Putting that reset inside the loop discards earlier additions. Keeping the reset and loop in the same cell also makes rerunning the cell produce a fresh total.

## 4. Count days with a recorded run

> **Concept · A condition inside a loop**
>
> The loop visits every distance. The `if` asks whether the current distance is greater than zero. We increase the counter only when that condition is true.


```python
run_days = 0  # Start a new count of days with a recorded run.
for distance_km in daily_distances_km:  # Visit every day's distance.
    if distance_km > 0:  # Check whether this day has a positive recorded distance.
        run_days = run_days + 1  # Count this day when the condition is true.
print("Days with a recorded run:", run_days)  # Display a count of five.
```

The `if` is indented inside the loop. The counter update is indented again inside the `if`. This is one loop containing one decision.

Predict the session numbers below. Will session 4 appear?

## 6. Reuse the weekly calculation

We can put a loop inside a function so the app can summarize different weeks. Recall that the parameter receives the supplied list and `return` sends the result back to the caller.


```python
def weekly_distance(distances_km):  # Define a function that accepts a list of daily distances.
    total_km = 0.0  # Start a fresh total for this function call.
    for distance_km in distances_km:  # Process each distance supplied to the function.
        total_km = total_km + distance_km  # Accumulate the distances one by one.
    return total_km  # Return the finished total after all items have been processed.
```

```python
this_week_km = weekly_distance(daily_distances_km)  # Calculate the total for the supplied week.
previous_distances_km = [2.0, 0.0, 4.0, 0.0, 3.0, 5.0, 0.0]  # Supply an independent previous week.
previous_week_km = weekly_distance(previous_distances_km)  # Reuse the function for the previous week.
print("This week's run distance:", this_week_km, "km")  # Display the current total of 20.0 km.
print("Previous week's run distance:", previous_week_km, "km")  # Display the previous total of 14.0 km.
```

> **Watch out · Finish the loop before returning**
>
> In `weekly_distance()`, `return` is inside the function but outside the loop. Returning from inside the loop would stop the function after the first processed item.

## 7. Predict and modify

Change Tuesday's distance, at index `1`, from 0.0 to 2.0 km. Predict the new total, then run the cell. The function recomputes the total from the current list each time.


```python
daily_distances_km[1] = 2.0  # Record a 2 km run on Tuesday.
updated_total_km = weekly_distance(daily_distances_km)  # Recalculate from the updated list.
print("Updated weekly distance:", updated_total_km, "km")  # Display the new total of 22.0 km.
```

### Key takeaways

- A `for` loop applies instructions to each list item.
- A total or counter starts before the loop and changes inside it.
- A condition can decide which items contribute to a count.
