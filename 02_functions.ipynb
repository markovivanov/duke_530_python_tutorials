## 1. Start with one run

> **The app calculation · Average pace**
>
> Pace is duration divided by distance. For a 5 km run taking 30 minutes, the average pace is 6 minutes per kilometer. All distances in today's examples are greater than zero.

We will use **decimal minutes per kilometer**. A pace of 6.5 means 6 minutes and 30 seconds per kilometer, because half a minute is 30 seconds.


```python
distance_km = 5.0  # Store the distance of the sample run.
duration_minutes = 30  # Store the time taken for the sample run.
pace = duration_minutes / distance_km  # Calculate average minutes per kilometer.
print("Average pace:", pace, "min/km")  # Display the calculated pace and its unit.
```

The result is **6.0 min/km**. We could type the same calculation for every run, but our app needs a reusable operation with clear inputs and an output.

## 2. Give the calculation a name

> **Concept · A function**
>
> A function is a named group of instructions. `def` introduces its definition. The names in parentheses are **parameters**: they represent the inputs the function will receive. A colon starts the body, and indentation groups the instructions inside it.

Our function has two inputs, `distance_km` and `duration_minutes`, and returns one result: pace. `return` sends that result back to the code that called the function.


```python
def calculate_pace(distance_km, duration_minutes):  # Define a reusable pace calculation with two inputs.
    pace = duration_minutes / distance_km  # Calculate minutes per kilometer for this call.
    return pace  # Send the calculated pace back to the caller.
```

Running this cell **defines** the function. It does not yet calculate a new run's pace, so no output is expected.

| Part | Meaning in our app |
|---|---|
| `calculate_pace` | The name of the reusable operation |
| `distance_km`, `duration_minutes` | Parameters that name the required inputs |
| Indented lines | Instructions run when we call the function |
| `return pace` | The result sent back to the caller |

## 3. Call the function

> **Concept · Arguments**
>
> A call uses the function's name followed by parentheses containing actual input values. Those supplied values are **arguments**. Here, the first argument is distance and the second is duration.


```python
sam_pace = calculate_pace(5.0, 30)  # Call the function for Sam's 5 km run lasting 30 minutes.
alex_pace = calculate_pace(4.0, 28)  # Reuse the function for Alex's 4 km run lasting 28 minutes.
print("Sam's pace:", sam_pace, "min/km")  # Display Sam's returned pace of 6.0.
print("Alex's pace:", alex_pace, "min/km")  # Display Alex's returned pace of 7.0.
```

> **Watch out · Input order matters**
>
> Our function expects distance first and duration second. Reversing them changes the calculation. Check the definition when you are unsure what to supply.

We can also pass variables as arguments. Each call receives the values those variables hold at that moment.


```python
new_distance_km = 3.0  # Store a new run's distance.
new_duration_minutes = 21  # Store the new run's duration.
new_pace = calculate_pace(new_distance_km, new_duration_minutes)  # Pass the stored values to the function.
print("New run pace:", new_pace, "min/km")  # Display the result of 7.0 minutes per kilometer.
```

## 4. Returning and displaying do different jobs

> **Concept · Keep the result available**
>
> `return` gives a value back to the caller, so the app can store or calculate with it. `print()` only displays information. A function with no explicit `return` gives back `None`, Python's value for “no value here.”

Compare this display-only function with `calculate_pace()`.


```python
def display_pace(distance_km, duration_minutes):  # Define a function that displays a pace.
    pace = duration_minutes / distance_km  # Calculate the pace inside the function.
    print("Displayed pace:", pace, "min/km")  # Show the value without returning it.

display_result = display_pace(5.0, 30)  # Run the display function and store its return value.
print("Value returned:", display_result)  # Show None because there was no explicit return.
```

Now use a returned pace in a second calculation. We can convert minutes per kilometer into seconds per kilometer.


```python
returned_pace = calculate_pace(5.0, 30)  # Store a numeric value returned by the pace function.
pace_seconds = returned_pace * 60  # Convert minutes per kilometer to seconds per kilometer.
print("Pace in seconds:", pace_seconds, "seconds/km")  # Display 360.0 seconds per kilometer.
```

## 5. Build a readable pace summary

Our original runner completed 5.4 km in 32 minutes. This division produces several decimal places. `round(value, places)` is a built-in function that rounds a number to a chosen number of decimal places.


```python
runner_name = "Sam"  # Store the name to show in the workout summary.
run_distance_km = 5.4  # Store the recorded run distance.
run_duration_minutes = 32  # Store the recorded run duration.
run_pace = calculate_pace(run_distance_km, run_duration_minutes)  # Reuse our function for this workout.
rounded_pace = round(run_pace, 2)  # Round the result to two decimal places for display.
print(runner_name, "ran at", rounded_pace, "min/km")  # Display a readable pace summary.
```

The displayed pace is **5.93 min/km**, expressed in decimal minutes.

> **Concept · Names inside a function**
>
> Parameters and variables assigned inside a function are local to that call. Our function uses the distance and duration we pass to it. Returning a result is how we make that result available outside the function.
