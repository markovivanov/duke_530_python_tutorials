## 1. Start with a familiar function

> **The app calculation · BMI**
>
> For this numeric profile feature, BMI is weight in kilograms divided by height in meters squared. “Squared” means multiplying the height by itself. We will calculate and display the number using fictional profiles.

The parentheses below group the two height values so Python multiplies them before dividing the weight. Our sample heights are all greater than zero.


```python
def calculate_bmi(weight_kg, height_m):  # Define a function that accepts weight and height.
    bmi = weight_kg / (height_m * height_m)  # Divide weight by height multiplied by itself.
    return bmi  # Return the calculated number.

sam_bmi = calculate_bmi(70.0, 1.75)  # Calculate BMI from Sam's fictional measurements.
print("Sam's BMI:", round(sam_bmi, 2))  # Display the result rounded to two decimal places.
```

The result is **22.86**. `round(value, 2)` rounds a number to two decimal places for display.

We can pass measurements to this function, but the app also needs to keep each person's measurements together. A profile gives us one object that holds related information and knows how to use it.

## 2. A profile holds related information

> **Concept · Class, object, attribute, method**
>
> A **class** defines a kind of object. An **object**, also called an **instance**, is one particular example of that class. An **attribute** is a value stored on an object. A **method** is a function defined inside a class.

| Idea | Fitness app example |
|---|---|
| Class | `FitnessProfile`, the definition of a profile |
| Object / instance | Sam's individual profile |
| Attribute | Sam's stored height |
| Method | A BMI calculation using Sam's stored measurements |

We will use one small class throughout the lesson. Read it once with the instructor, then return to the explanation below. You do not need to memorize it.


```python
class FitnessProfile:  # Define the kind of profile our app can create.
    def __init__(self, name, height_m, weight_kg):  # Initialize each new profile with its own input values.
        self.name = name  # Store this profile's name.
        self.height_m = height_m  # Store this profile's height in meters.
        self.weight_kg = weight_kg  # Store this profile's weight in kilograms.

    def calculate_bmi(self):  # Define a method that uses the current profile's data.
        bmi = self.weight_kg / (self.height_m * self.height_m)  # Calculate BMI from stored measurements.
        return bmi  # Return the result to the caller.
```

### Read the class one part at a time

- `class FitnessProfile:` names the class. We use capitals to distinguish the class name from variable names.
- `__init__` has **two underscores before and after** `init`. Python calls this initialization method when we create an instance.
- `self` refers to the particular object being used. `self.height_m` means “the height stored on this profile.”
- `self.name = name` stores the supplied `name` value as an attribute of the new object.
- `calculate_bmi` is a method because its definition sits inside the class. It uses the current object's attributes and returns a result.

> **Watch out · Two indentation levels**
>
> Method definitions are indented four spaces inside the class. The instructions inside each method are indented another four spaces. These levels tell Python which instructions belong together.

## 3. Create one profile

Calling the class name creates an object. We supply name, height, and weight in the same order as the parameters after `self`. Python supplies `self` automatically.


```python
sam_profile = FitnessProfile("Sam", 1.75, 70.0)  # Create Sam's profile with height in meters and weight in kilograms.
print("Name:", sam_profile.name)  # Read the name attribute from Sam's profile.
print("Height:", sam_profile.height_m, "m")  # Read and display Sam's stored height.
print("Weight:", sam_profile.weight_kg, "kg")  # Read and display Sam's stored weight.
```

> **Concept · Dot notation**
>
> A dot connects an object to one of its attributes or methods. `sam_profile.name` retrieves an attribute. `sam_profile.calculate_bmi()` calls a method; the parentheses ask it to run.

Notice that we do not repeat the measurements in the method call. They are already stored in the object.


```python
sam_bmi = sam_profile.calculate_bmi()  # Call the method using Sam's stored measurements.
print("Sam's BMI:", round(sam_bmi, 2))  # Display the returned result of 22.86.
```

## 4. Create a second profile

Both objects follow the same class definition, but each stores its own values. Predict Alex's name and weight before running the cell.


```python
alex_profile = FitnessProfile("Alex", 1.60, 64.0)  # Create a separate profile for Alex.
print("Name:", alex_profile.name)  # Display the name on Alex's profile.
print("Weight:", alex_profile.weight_kg, "kg")  # Display Alex's stored weight.
alex_bmi = alex_profile.calculate_bmi()  # Run the same method using Alex's measurements.
print("Alex's BMI:", round(alex_bmi, 2))  # Display Alex's result of 25.0.
```

## 5. Update one profile

Our app can replace a recorded value by assigning to an attribute. In this example, we correct Sam's sample weight to 72 kg. What should happen to Alex's stored weight?


```python
sam_profile.weight_kg = 72.0  # Replace the weight stored on Sam's profile.
updated_sam_bmi = sam_profile.calculate_bmi()  # Calculate again using Sam's updated measurement.
print("Sam's updated BMI:", round(updated_sam_bmi, 2))  # Display the recalculated result of 23.51.
print("Alex's weight:", alex_profile.weight_kg, "kg")  # Confirm that Alex's separate value is still 64.0.
```

The method reads the object's values each time it runs. The earlier variable `sam_bmi` still holds the result of an earlier call; we stored the new result in `updated_sam_bmi`.

## 6. Build a profile summary

We now have enough to display a small profile card as text. Select a profile, read its attributes, and call its method.


```python
print("Fitness profile:", sam_profile.name)  # Start the summary with the person's name.
print("Height:", sam_profile.height_m, "m")  # Display the height with its unit.
print("Weight:", sam_profile.weight_kg, "kg")  # Display the current stored weight.
current_bmi = sam_profile.calculate_bmi()  # Calculate the profile's current BMI value.
print("BMI:", round(current_bmi, 2))  # Display the rounded numeric result.
```
