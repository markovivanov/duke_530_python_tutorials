## 1. Compare a step count with a goal

> **Concept · Comparisons produce Boolean values**
>
> A comparison asks a question about two values and returns `True` or `False`. `>=` asks whether the left value is greater than or equal to the right value.


```python
steps_today = 6200  # Store the fictional runner's current step count.
step_goal = 6000  # Store the goal chosen for this example.
goal_reached = steps_today >= step_goal  # Check whether the count is at least the goal.
print("Goal reached:", goal_reached)  # Display True because 6200 is at least 6000.
```

These operators let the app ask slightly different questions.

| Operator | Question | Fitness use |
|---|---|---|
| `>` | Greater than? | Did the count exceed the goal? |
| `<` | Less than? | Is the count below the goal? |
| `>=` | Greater than or equal to? | Has the goal been reached? |
| `<=` | Less than or equal to? | Is the count at or below the goal? |
| `==` | Equal to? | Does the count exactly match the goal? |
| `!=` | Not equal to? | Does the count differ from the goal? |

Predict these results using the stored count of 6200 and goal of 6000.


```python
print(steps_today > step_goal)  # Check whether the runner is above the goal.
print(steps_today == step_goal)  # Check whether the count exactly matches the goal.
print(steps_today < step_goal)  # Check whether the count is below the goal.
```

## 2. Run an instruction only when a condition is true

> **Concept · if**
>
> `if` starts a conditional statement. Python checks the condition before the colon. If it is `True`, Python runs the indented body. If it is `False`, Python skips that body.


```python
if steps_today >= step_goal:  # Check whether the selected goal has been reached.
    print("Step goal reached!")  # Display this message only when the condition is true.
print("Step count checked.")  # Always display this message because it is outside the if body.
```

The last line is unindented, so it runs regardless of the condition. Change `steps_today` to 4000 in the first cell, rerun that cell, then rerun this decision. Which message disappears?

## 3. Choose between two messages

> **Concept · else**
>
> `else` provides the alternative when the `if` condition is false. It has a colon but no condition of its own. Exactly one of the two bodies runs.


```python
steps_today = 4500  # Set a fresh count for a below-goal example.
step_goal = 6000  # Set the example's chosen goal.
if steps_today >= step_goal:  # Check whether the goal has been reached.
    print("Step goal reached!")  # Display the reached-goal message when the condition is true.
else:  # Handle the case where the count is below the goal.
    steps_remaining = step_goal - steps_today  # Calculate the difference from the goal.
    print("Steps remaining to the chosen goal:", steps_remaining)  # Display the remaining count.
```

The result is **1500 steps remaining**. We calculate the difference inside the `else` body because that is the case where we need it.

## 4. Distinguish below, at, and above

> **Concept · elif**
>
> `elif` means “else if.” Python checks conditions from top to bottom and runs the first body whose condition is true. The final `else` covers anything left over.

Our app will show one of three messages. Start with the count exactly equal to the goal.


```python
steps_today = 6000  # Choose a count exactly equal to the goal.
step_goal = 6000  # Store the example goal.
if steps_today > step_goal:  # First check whether the count is above the goal.
    print("You are above your chosen step goal.")  # Display the above-goal message.
elif steps_today == step_goal:  # Otherwise check for an exact match.
    print("You have exactly reached your step goal.")  # Display the exact-match message.
else:  # Handle the remaining case where the count is below the goal.
    print("You are below your chosen step goal.")  # Display the below-goal message.
```

> **Watch out · Assignment and equality**
>
> `step_goal = 6000` assigns a value. `steps_today == step_goal` compares two values. A comparison in these conditions needs `==`, not `=`.

## 5. Put the decision in a reusable function

Our app needs this message whenever a step count changes. Recall that `def` defines a function, parameters receive inputs, and `return` sends a value back to its caller.


```python
def step_goal_message(steps, goal):  # Define a function that accepts a count and a chosen goal.
    if steps > goal:  # Check whether this count exceeds this goal.
        return "Above goal"  # Return a label for an above-goal count.
    elif steps == goal:  # Check for an exact match if the first condition was false.
        return "Goal reached"  # Return a label for an exact match.
    else:  # Handle a count below the goal.
        return "Below goal"  # Return a label for the remaining case.
```

```python
current_steps = 5200  # Store a fictional user's current count.
chosen_goal = 6000  # Store the goal selected for this summary.
message = step_goal_message(current_steps, chosen_goal)  # Get the appropriate message from the function.
print("Today's steps:", current_steps)  # Display the recorded count.
print("Chosen goal:", chosen_goal)  # Display the goal used for comparison.
print("Progress:", message)  # Display the returned progress label.
```

## 6. Check all three possibilities

Each comparison should lead to the intended message. Predict the three outputs before running the cell.


```python
print(step_goal_message(4500, 6000))  # Check a count below the chosen goal.
print(step_goal_message(6000, 6000))  # Check a count exactly equal to the goal.
print(step_goal_message(7200, 6000))  # Check a count above the goal.
```

### Predict, change, and discuss

Keep a count of 6000 but change the chosen goal to 7000. Which label should the function return? Then try a goal of 5000. The same recorded activity can produce different progress messages when the chosen goal changes.

> **Watch out · A comparison needs appropriate values**
>
> Keep both the step count and goal as numbers. A value such as `"6000"` is text because it has quotes. Comparing text with a number using `>` raises a `TypeError`; inspect the values and their types.

- An `if` / `elif` / `else` chain chooses one branch.
